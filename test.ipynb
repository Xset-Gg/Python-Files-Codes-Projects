```python
import sys
from PyQt5.QtWidgets import QApplication, QWidget, QLabel, QVBoxLayout
from PyQt5.QtCore import QTimer, QTime, Qt


class MainWindow(QWidget):
    def __init__(self):
        super().__init__()

        self.setWindowTitle("Digital Clock")
        self.setGeometry(700, 300, 400, 200)

        self.label = QLabel(self)

        self.label.setAlignment(Qt.AlignCenter)

        self.label.setStyleSheet("""
            font-size: 40px;
            font-family: Arial;
        """)

        layout = QVBoxLayout()
        layout.addWidget(self.label)

        self.setLayout(layout)

        # Timer
        self.timer = QTimer()
        self.timer.timeout.connect(self.update_time)
        self.timer.start(1000)

        self.update_time()

    def update_time(self):
        current_time = QTime.currentTime().toString("hh:mm:ss")
        self.label.setText(current_time)


if __name__ == "__main__":
    app = QApplication(sys.argv)

    window = MainWindow()
    window.show()

    sys.exit(app.exec())
```

```python
import sys
import requests
from datetime import datetime

from PyQt5.QtWidgets import (
    QApplication, QWidget, QLabel, QPushButton, 
    QLineEdit, QVBoxLayout, QHBoxLayout, 
    QComboBox, QFrame, QGridLayout, QMessageBox
)
from PyQt5.QtCore import Qt
from PyQt5.QtGui import QFont


class WeatherApp(QWidget):

    def __init__(self):
        super().__init__()
        self.init_ui()
        self.apply_styles()

    def init_ui(self):
        self.setWindowTitle("🇵🇰 Pakistan Weather App")
        self.setGeometry(500, 200, 450, 600)

        # ========== HEADER ==========
        self.header = QLabel("🌤 Pakistan Weather")
        self.header.setAlignment(Qt.AlignCenter)
        self.header.setFont(QFont("Segoe UI", 22, QFont.Bold))

        self.subheader = QLabel("Real-time Weather Updates")
        self.subheader.setAlignment(Qt.AlignCenter)

        # ========== CITY INPUT SECTION ==========
        self.city_combo = QComboBox()
        self.city_combo.setEditable(True)
        self.city_combo.setPlaceholderText("Or type city name...")
        
        # Popular Pakistani cities
        cities = [
            "Karachi", "Lahore", "Islamabad", "Rawalpindi", 
            "Faisalabad", "Peshawar", "Multan", "Gujranwala", 
            "Quetta", "Sialkot", "Bahawalpur", "Sargodha", 
            "Sheikhupura", "Gujrat", "Larkana", "Sukkur"
        ]
        self.city_combo.addItems(cities)

        self.search_btn = QPushButton("🔍 Get Weather")
        self.search_btn.setCursor(Qt.PointingHandCursor)
        self.search_btn.clicked.connect(self.get_weather)

        # Input layout
        input_layout = QHBoxLayout()
        input_layout.addWidget(self.city_combo)
        input_layout.addWidget(self.search_btn)

        # ========== LOADING LABEL ==========
        self.loading = QLabel("")
        self.loading.setAlignment(Qt.AlignCenter)

        # ========== MAIN RESULT CARD ==========
        self.result_card = QFrame()
        self.result_card.setObjectName("resultCard")
        self.result_card.setFrameShape(QFrame.StyledPanel)
        
        self.result_label = QLabel("Enter a city to get weather info")
        self.result_label.setAlignment(Qt.AlignCenter)
        self.result_label.setWordWrap(True)
        self.result_label.setFont(QFont("Segoe UI", 11))

        card_layout = QVBoxLayout()
        card_layout.addWidget(self.result_label)
        self.result_card.setLayout(card_layout)

        # ========== DETAILS GRID ==========
        self.details_grid = QGridLayout()
        self.details_grid.setSpacing(15)

        # Detail boxes
        self.temp_box = self.create_detail_box("🌡 Temperature", "--")
        self.feels_box = self.create_detail_box("🤔 Feels Like", "--")
        self.humidity_box = self.create_detail_box("💧 Humidity", "--")
        self.wind_box = self.create_detail_box("💨 Wind", "--")
        self.pressure_box = self.create_detail_box("📊 Pressure", "--")
        self.visibility_box = self.create_detail_box("👁 Visibility", "--")

        self.details_grid.addWidget(self.temp_box, 0, 0)
        self.details_grid.addWidget(self.feels_box, 0, 1)
        self.details_grid.addWidget(self.humidity_box, 1, 0)
        self.details_grid.addWidget(self.wind_box, 1, 1)
        self.details_grid.addWidget(self.pressure_box, 2, 0)
        self.details_grid.addWidget(self.visibility_box, 2, 1)

        # ========== FOOTER ==========
        self.footer = QLabel("Powered by OpenWeatherMap API")
        self.footer.setAlignment(Qt.AlignCenter)

        # ========== MAIN LAYOUT ==========
        main_layout = QVBoxLayout()
        main_layout.setSpacing(15)
        main_layout.addWidget(self.header)
        main_layout.addWidget(self.subheader)
        main_layout.addLayout(input_layout)
        main_layout.addWidget(self.loading)
        main_layout.addWidget(self.result_card)
        main_layout.addLayout(self.details_grid)
        main_layout.addStretch()
        main_layout.addWidget(self.footer)

        self.setLayout(main_layout)

    def create_detail_box(self, title, value):
        """Create a styled detail card"""
        frame = QFrame()
        frame.setObjectName("detailBox")
        frame.setFrameShape(QFrame.StyledPanel)
        
        title_lbl = QLabel(title)
        title_lbl.setObjectName("detailTitle")
        title_lbl.setAlignment(Qt.AlignCenter)
        
        value_lbl = QLabel(value)
        value_lbl.setObjectName("detailValue")
        value_lbl.setAlignment(Qt.AlignCenter)
        value_lbl.setFont(QFont("Segoe UI", 14, QFont.Bold))
        
        layout = QVBoxLayout()
        layout.addWidget(title_lbl)
        layout.addWidget(value_lbl)
        frame.setLayout(layout)
        
        # Store reference to update later
        frame.value_label = value_lbl
        return frame

    def apply_styles(self):
        self.setStyleSheet("""
            QWidget {
                background-color: #1a1a2e;
                color: #eee;
                font-family: 'Segoe UI', Arial;
            }
            
            QLabel#detailTitle {
                color: #a0a0a0;
                font-size: 12px;
            }
            
            QLabel#detailValue {
                color: #00d4aa;
                font-size: 16px;
            }
            
            QFrame#resultCard {
                background-color: #16213e;
                border-radius: 15px;
                padding: 20px;
                border: 1px solid #0f3460;
            }
            
            QFrame#detailBox {
                background-color: #16213e;
                border-radius: 10px;
                padding: 15px;
                border: 1px solid #0f3460;
            }
            
            QComboBox {
                background-color: #16213e;
                color: white;
                border: 2px solid #0f3460;
                border-radius: 8px;
                padding: 10px;
                font-size: 14px;
            }
            
            QComboBox::drop-down {
                border: none;
                padding-right: 10px;
            }
            
            QComboBox QAbstractItemView {
                background-color: #16213e;
                color: white;
                selection-background-color: #0f3460;
            }
            
            QPushButton {
                background-color: #e94560;
                color: white;
                border: none;
                border-radius: 8px;
                padding: 10px 20px;
                font-size: 14px;
                font-weight: bold;
            }
            
            QPushButton:hover {
                background-color: #ff6b6b;
            }
            
            QPushButton:pressed {
                background-color: #c73e54;
            }
            
            QLineEdit {
                background-color: #16213e;
                color: white;
                border: 2px solid #0f3460;
                border-radius: 8px;
                padding: 10px;
                font-size: 14px;
            }
        """)

    def get_weather(self):
        city = self.city_combo.currentText().strip()
        
        if not city:
            QMessageBox.warning(self, "Missing Input", "Please enter or select a city!")
            return

        # Show loading
        self.loading.setText("⏳ Fetching weather data...")
        self.loading.setStyleSheet("color: #ffd700; font-size: 14px;")
        QApplication.processEvents()

        api_key = "7fb8e04e9652b878c23771b5eb84d3ba"
        url = (
            f"https://api.openweathermap.org/data/2.5/weather"
            f"?q={city},PK&appid={api_key}&units=metric"
        )

        try:
            response = requests.get(url, timeout=10)
            data = response.json()

            if data.get("cod") == 200:
                self.display_weather(data, city)
            else:
                self.result_label.setText(f"❌ City '{city}' not found!\nPlease check spelling.")
                self.clear_details()
                QMessageBox.critical(self, "Error", f"City '{city}' not found in Pakistan!")

        except requests.exceptions.ConnectionError:
            self.result_label.setText("❌ No Internet Connection!")
            self.clear_details()
            QMessageBox.critical(self, "Network Error", "Please check your internet connection.")
        
        except requests.exceptions.Timeout:
            self.result_label.setText("⏱ Request Timed Out!")
            self.clear_details()
            QMessageBox.critical(self, "Timeout", "Server is taking too long to respond.")
        
        except Exception as e:
            self.result_label.setText(f"❌ Error: {str(e)}")
            self.clear_details()
            QMessageBox.critical(self, "Error", f"Something went wrong:\n{str(e)}")
        
        finally:
            self.loading.setText("")

    def display_weather(self, data, city):
        # Extract data
        main = data["main"]
        weather = data["weather"][0]
        wind = data["wind"]
        
        temp = main["temp"]
        feels_like = main["feels_like"]
        humidity = main["humidity"]
        pressure = main["pressure"]
        condition = weather["description"].title()
        wind_speed = wind["speed"]
        visibility = data.get("visibility", 0) / 1000  # Convert to km
        
        # Weather emoji based on condition
        emoji = self.get_weather_emoji(condition)
        
        # Update main card
        self.result_label.setText(
            f"<h2>{emoji} {city.title()}</h2>"
            f"<h1 style='color:#00d4aa;'>{temp}°C</h1>"
            f"<p style='font-size:16px;'>{condition}</p>"
        )

        # Update detail boxes
        self.temp_box.value_label.setText(f"{temp} °C")
        self.feels_box.value_label.setText(f"{feels_like} °C")
        self.humidity_box.value_label.setText(f"{humidity} %")
        self.wind_box.value_label.setText(f"{wind_speed} m/s")
        self.pressure_box.value_label.setText(f"{pressure} hPa")
        self.visibility_box.value_label.setText(f"{visibility:.1f} km")

    def get_weather_emoji(self, condition):
        condition = condition.lower()
        if "clear" in condition or "sun" in condition:
            return "☀️"
        elif "cloud" in condition:
            return "☁️"
        elif "rain" in condition or "drizzle" in condition:
            return "🌧"
        elif "thunder" in condition or "storm" in condition:
            return "⛈"
        elif "snow" in condition:
            return "❄️"
        elif "mist" in condition or "fog" in condition or "haze" in condition:
            return "🌫"
        else:
            return "🌤"

    def clear_details(self):
        for box in [self.temp_box, self.feels_box, self.humidity_box, 
                    self.wind_box, self.pressure_box, self.visibility_box]:
            box.value_label.setText("--")


if __name__ == "__main__":
    app = QApplication(sys.argv)
    window = WeatherApp()
    window.show()
    sys.exit(app.exec_())
```
