```python
def count_up_to(n):
  count = 1
  while count <= n:
    yield count
    count += 1

for num in count_up_to(5):
  print(num)
```

```python
def large_sequence(n):
  for i in range(n):
    yield i


gen = large_sequence(1000000)
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def fibonacci():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a + b


gen = fibonacci()
for _ in range(6):
  print(next(gen))
```

```python
def abu(a, b, ali):
    if ali == "+":
        return a + b
    elif ali == "-":
        return a - b
    elif ali == "*":
        return a * b
    elif ali == "/":
        return a / b
    else:
        return "Invalid Operator"

print(abu(20, 14, "+"))
print(abu(20, 19, "*"))
```
